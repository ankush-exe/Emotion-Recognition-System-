```python
from pathlib import Path
from PIL import Image
from collections import Counter
import yaml
```

```python
dataset_path = Path("../dataset/YOLO_format")

train_images = dataset_path / "train" / "images"
train_labels = dataset_path / "train" / "labels"

valid_images = dataset_path / "valid" / "images"
valid_labels = dataset_path / "valid" / "labels"
```

```python
print("========== DATASET STRUCTURE ==========\n")

print("Dataset Path:", dataset_path)

print("\nTraining")
print("Images :", train_images)
print("Labels :", train_labels)

print("\nValidation")
print("Images :", valid_images)
print("Labels :", valid_labels)

print("\nExists?")
print(train_images.exists())
print(train_labels.exists())
print(valid_images.exists())
print(valid_labels.exists())
```

```python
train_imgs = list(train_images.glob("*"))
valid_imgs = list(valid_images.glob("*"))

print("Training Images :", len(train_imgs))
print("Validation Images :", len(valid_imgs))
print("Total Images :", len(train_imgs)+len(valid_imgs))
```

```python
sample = train_imgs[0]

img = Image.open(sample)

print("Image Name :", sample.name)
print("Format :", img.format)
print("Width :", img.width)
print("Height :", img.height)
print("Mode :", img.mode)

channels = len(img.getbands())

print("Channels :", channels)
```

```python
label_files = list(train_labels.glob("*.txt")) + list(valid_labels.glob("*.txt"))

print("Number of Label Files :", len(label_files))
```

```python
annotation_count = 0
class_counter = Counter()

for file in label_files:

    with open(file) as f:
        lines = f.readlines()

    annotation_count += len(lines)

    for line in lines:
        class_id = int(line.split()[0])
        class_counter[class_id] += 1

print("Total Annotations :", annotation_count)

print("\nClass IDs")

for k,v in sorted(class_counter.items()):
    print(f"Class {k}: {v}")
```

```python
yaml_file = dataset_path / "data.yaml"

with open(yaml_file,"r") as f:
    data = yaml.safe_load(f)

print(data)
```

```python
print("\nEmotion Mapping\n")

for idx,name in enumerate(data["names"]):
    print(f"{idx} -> {name}")
```

```python
print("========== DATASET SUMMARY ==========")

print(f"Training Images : {len(train_imgs)}")
print(f"Validation Images : {len(valid_imgs)}")
print(f"Total Images : {len(train_imgs)+len(valid_imgs)}")

print(f"Label Files : {len(label_files)}")
print(f"Annotations : {annotation_count}")

print("\nEmotion Classes")

if isinstance(data["names"], dict):
    for k,v in data["names"].items():
        print(f"{k} -> {v}")
else:
    for i,name in enumerate(data["names"]):
        print(f"{i} -> {name}")
```
